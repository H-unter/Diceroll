```python
from Spell import Spell
import matplotlib.pyplot as plt
```

# Level 1 Spells

```python
burning_hands = Spell(
    name="Burning Hands",
    starting_level=1,
    starting_damage_diceroll="3d6",
    damage_increment_diceroll="1d6",
    num_levels_per_damage_increase=1,
    is_saving_throw=True
)
burning_hands.plot_damage()
plt.savefig("spell_plots/burning_hands.png", dpi=500, bbox_inches='tight')
```

```python
guiding_bolt = Spell(
    name="Guiding Bolt",
    starting_level=1,
    starting_damage_diceroll="4d6",
    damage_increment_diceroll="1d6",
    num_levels_per_damage_increase=1,
    is_saving_throw=False
)
guiding_bolt.plot_damage()
plt.savefig("spell_plots/guiding_bolt.png", dpi=500, bbox_inches='tight')
```

```python
healing_word = Spell(
    name="Healing Word",
    starting_level=1,
    starting_damage_diceroll="1d4+5",
    damage_increment_diceroll="1d4",
    num_levels_per_damage_increase=1,
    is_saving_throw=False,
    is_healing=True
)
healing_word.plot_damage()
plt.savefig("spell_plots/healing_word.png", dpi=500, bbox_inches='tight')
```

```python
inflict_wounds = Spell(
    name="Inflict Wounds",
    starting_level=1,
    starting_damage_diceroll="3d10",
    damage_increment_diceroll="1d10",
    num_levels_per_damage_increase=1,
    is_saving_throw=False
)
inflict_wounds.plot_damage()
inflict_wounds.boxplot_damage()
# plt.savefig("spell_plots/inflict_wounds.png", dpi=500, bbox_inches='tight')
```
