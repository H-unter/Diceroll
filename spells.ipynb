```python
from Spell import Spell
import matplotlib.pyplot as plt
```

```python
burning_hands = Spell(
    name="Burning Hands",
    starting_level=1,
    starting_damage_diceroll="3d6",
    damage_increment_diceroll="1d6",
    num_levels_per_damage_increase=1,
    is_saving_throw=True
)
burning_hands.plot_damage(max_level=4)
plt.savefig("spell_plots/burning_hands.png", dpi=500, bbox_inches='tight')
```

```python
guiding_bolt = Spell(
    name="Guiding Bolt",
    starting_level=1,
    starting_damage_diceroll="4d6",
    damage_increment_diceroll="1d6",
    num_levels_per_damage_increase=1,
    is_saving_throw=False
)
guiding_bolt.plot_damage(max_level=4)
plt.savefig("spell_plots/guiding_bolt.png", dpi=500, bbox_inches='tight')
```

```python
spirit_guardians = Spell(
    name="Spirit Guardians",
    starting_level=3,
    starting_damage_diceroll="3d8",
    damage_increment_diceroll="1d8",
    num_levels_per_damage_increase=1,
    is_saving_throw=True
)
spirit_guardians.plot_damage(max_level=9, colormap=plt.cm.plasma)
plt.savefig("spell_plots/spirit_guardians.png", dpi=500, bbox_inches='tight')
```
