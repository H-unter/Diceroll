# Cantrips

```python

sacred_flame_caster_level_to_dice_prompt = {
    1: "1d8",
    5: "2d8",
    11: "3d8",
    17: "4d8"
}
sacred_flame = dice_roll_toolbox(sacred_flame_caster_level_to_dice_prompt)
sacred_flame.plot_values(title=f"Sacred Flame Damage Distribution", upgrade="Caster Level") # Slot
```
