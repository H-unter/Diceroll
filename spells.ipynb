```python
import matplotlib.pyplot as plt
from dice_roll_toolbox import Spell  # Import the Spell class
cleric_level = 4
wisdom_modifier = 4
strength_modifier = 2
```

# Channel Divinity: Radiance of The Dawn

```python
radiance_of_dawn = f"2d10+{cleric_level}"
radiance_of_dawn_damage = Spell(radiance_of_dawn)
radiance_of_dawn_damage.plot_values(name=f"Radiance of Dawn, Cleric Level {cleric_level}", x_tick_interval=1)
```
