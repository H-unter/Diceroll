```python
from Spell import Spell
import matplotlib.pyplot as plt
```

```python
burning_hands = Spell(
    name="Burning Hands",
    starting_level=1,
    starting_damage_diceroll="3d6",
    damage_increment_diceroll="1d6",
    num_levels_per_damage_increase=1,
    is_halved_on_miss=True
)
burning_hands.plot_damage(max_level=4)
```

```python
guiding_bolt = Spell(
    name="Guiding Bolt",
    starting_level=1,
    starting_damage_diceroll="4d6",
    damage_increment_diceroll="1d6",
    num_levels_per_damage_increase=1,
    is_halved_on_miss=False
)
guiding_bolt.plot_damage(max_level=4)
```

```python
inflict_wounds = Spell(
    name="Inflict Wounds",
    starting_level=1,
    starting_damage_diceroll="3d10",
    damage_increment_diceroll="1d10",
    num_levels_per_damage_increase=1,
    is_halved_on_miss=False
)
inflict_wounds.plot_damage(max_level=4)
```

```python
spirit_guardians = Spell(
    name="Spirit Guardians",
    starting_level=3,
    starting_damage_diceroll="3d8",
    damage_increment_diceroll="1d8",
    num_levels_per_damage_increase=1,
    is_halved_on_miss=True
)
spirit_guardians.plot_damage(max_level=6, colormap=plt.cm.plasma)
```
