```python
from Spell import Spell
import matplotlib.pyplot as plt
```

# Level 0 Spells

```python
true_strike = Spell(
    name="True Strike",
    starting_level=0,
    hardcoded_level_to_diceroll={
        5: "1d6",
        11: "2d6",
        17: "3d6",
    }
)
true_strike.plot_damage()
true_strike.boxplot_damage(colormap = plt.cm.Blues)
plt.savefig("spell_plots/true_strike.png", dpi=500, bbox_inches='tight')
```
